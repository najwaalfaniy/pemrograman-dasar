```python
#Biodata Saya

nama = "Najwa"
umur = 21
jurusan = "Elektro"
prodi = "TRPE"
semester = 3
matakuliah = 'Pemrograman Dasar'
mahasiswa_aktif = True
hobi = ("berenang, badminton")
warna_favorite = ("abu", "biru", "coklat")
tinggi = 160.2
alamat = None

#perintah untuk menampilkan nilai pada variabel

print(f"Halo, nama saya {nama}.")
print(f"Saya berusia {umur} tahun.")
print(f"Saya mahasiswa jurusan {jurusan}.")
print(f"Saya sedang berkuliah di prodi {prodi}.")
print(f"Saya sedang berada di semester {semester}.")
print(f"Saya sedang mempelajari mata kuliah {matakuliah}.")
print(f"Status saya adalah mahasiswa aktif {mahasiswa_aktif}")
print(f"Hobi saya adalah {hobi}.")
print(f"Warna favorite saya adalah {warna_favorite}")
print(f"Tinggi badan saya adalah {tinggi} cm.")
print(f"Alamat saya adalah {alamat}.")


```
