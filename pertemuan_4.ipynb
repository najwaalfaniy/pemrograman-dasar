4. Modifikasi agar variable
nilai_mahasiswa dan
standard_kelulusan bisa menerima
inputan melalui keyboard dan Modifikasi agas nilai mahasiswa
terdiri dari 5 variable dengan input via
keyboard


```python
# Nama      : Najwa Alfaniy
# NIM       : 4232501032

# Menghitung nilai akhir mahasiswa

# Input 5 komponen nilai melalui keyboard
tugas = float(input("Masukkan nilai Tugas: "))
kuis = float(input("Masukkan nilai Kuis: "))
uts = float(input("Masukkan nilai UTS: "))
uas = float(input("Masukkan nilai UAS: "))
praktikum = float(input("Masukkan nilai Praktikum: "))

# Menghitung nilai akhir mahasiswa
nilai_mahasiswa = (
    (0.20 * tugas) +
    (0.10 * kuis) +
    (0.25 * uts) +
    (0.30 * uas) +
    (0.15 * praktikum)
)

# Input standar kelulusan melalui keyboard
standar_kelulusan = float(input("Masukkan standar kelulusan: "))

# Menentukan status mahasiswa
apakah_lulus = nilai_mahasiswa >= standar_kelulusan
perlu_remedial = nilai_mahasiswa < standar_kelulusan

# Menampilkan hasil
print("\n=== HASIL PENILAIAN ===")
print("Nilai akhir mahasiswa:", nilai_mahasiswa)
print("Standar kelulusan:", standar_kelulusan)
print("Apakah Mahasiswa Lulus?", apakah_lulus)
print("Perlu Remedial?", perlu_remedial)
```

5. Modifikasi inputan via keyboard, modifikasi agar total diskon yang di berikan menjadi 35% dengan belanja di atas 1000000

```python
# Nama      : Najwa Alfaniy
# NIM       : 4232501032

belanja = int(input("Masukkan total belanja: "))

diskon = (0, 0.35 * belanja)[belanja > 1000000]

total_harga = belanja - diskon

print(f"Total belanja : Rp{belanja:,.0f}")
print(f"Diskon        : Rp{diskon:,.0f}")
print(f"Total harga   : Rp{total_harga:,.0f}")
```

6. Menentukan Tahun Kabisat

Sebuah tahun disebut tahun kabisat jika memenuhi salah satu dari dua kondisi berikut:
a. Tahun tersebut habis dibagi 4, tetapi tidak habis dibagi 100.
b. Tahun tersebut habis dibagi 400.

Buatlah sebuah ekspresi logika kompleks untuk menentukan apakan suatu tahun merupakan tahun kabisat.

```python
# Nama      : Najwa Alfaniy
# NIM       : 4232501032

tahun = int(input("Masukkan tahun: "))

tahun_kabisat = (tahun % 400 == 0) or (
    tahun % 4 == 0 and tahun % 100 != 0
)

print(f"Tahun {tahun} merupakan tahun kabisat: {tahun_kabisat}")
```

7. Buatlah program sederhana versi anda masing-masing dengan lingkup, fungsi input, fungsi output dan operasi
artimatika!

```python
# Nama      : Najwa Alfaniy
# NIM       : 4232501032

def hitung_energi():
    daya = float(input("Masukkan daya listrik (W): "))
    waktu = float(input("Masukkan waktu penggunaan (jam): "))

    energi = daya * waktu

    print("\n=== HASIL PERHITUNGAN ===")
    print(f"Daya       : {daya:.2f} W")
    print(f"Waktu      : {waktu:.2f} jam")
    print(f"Energi     : {energi:.2f} Wh")


# Pemanggilan fungsi
hitung_energi()
```
