*Python mengenal beberapa jenis ekspresi (aritmatika, relasional, logika, assignment). Carilah referensi (dokumentasi resmi Python atau sumber lain) yang menjelaskan urutan prioritas operator (operator precedence) di Python, lalu jelaskan dengan kata-katamu sendiri: mengapa ekspresi 10 + 5 * 2 menghasilkan 20, bukan 30? Sertakan sumber yang kamu gunakan.*



Operator precedence adalah aturan yang menentukan operator mana yang dikerjakan terlebih dahulu ketika dalam satu ekspresi terdapat beberapa operator. Python memiliki tingkatan prioritas operator. Operator dengan prioritas lebih tinggi akan diproses lebih dahulu. Tanda kurung dapat digunakan untuk mengubah atau memperjelas urutan pengerjaan suatu ekspresi. Dokumentasi resmi Python mencantumkan bahwa operator perkalian *, pembagian /, pembagian bulat //, dan modulus % memiliki prioritas lebih tinggi daripada penjumlahan + dan pengurangan -.

Contoh:

hasil = 10 + 5 * 2
print(hasil)

Output: 20

Python mengerjakan perkalian terlebih dahulu:

5 * 2 = 10
10 + 10 = 20

Jadi, hasilnya adalah 20, bukan 30.

Jika ingin penjumlahan dilakukan terlebih dahulu, tanda kurung dapat digunakan:

hasil = (10 + 5) * 2
print(hasil)

Output: 30

- operator precedence penting karena menentukan urutan operasi dalam sebuah ekspresi.
- Tanda kurung dapat digunakan ketika kita ingin menentukan urutan perhitungan secara khusus atau membuat kode lebih mudah dipahami.

Contoh operator relasional:

(>) → lebih besar dari

(<) → lebih kecil dari

(==) → sama dengan

'or' → True jika setidaknya salah satu kondisi bernilai True

'not' → membalik nilai Boolean

merupakan operator relasional karena digunakan untuk membandingkan nilai. and merupakan operator logika yang menggabungkan kedua kondisi tersebut.

- Perbedaan mendasarnya adalah operator relasional digunakan untuk melakukan perbandingan

*Buatlah program Python yang meminta pengguna memasukkan dua bilangan bulat, lalu tampilkan hasil dari operasi +, -, *, /, // (pembagian bulat), dan % (modulus) antara kedua bilangan tersebut.*


```python
# Nama      : Najwa Alfaniy
# NIM       : 4232501032
# Tanggal   : 23 September 2026
# Tujuan    : Menghitung operasi aritmatika antara dua bilangan bulat.

bilangan1 = int(input("Masukkan bilangan pertama: "))
bilangan2 = int(input("Masukkan bilangan kedua: "))

print(f"Hasil penjumlahan (+)  : {bilangan1 + bilangan2}")
print(f"Hasil pengurangan (-)  : {bilangan1 - bilangan2}")
print(f"Hasil perkalian (*)    : {bilangan1 * bilangan2}")
print(f"Hasil pembagian (/)    : {bilangan1 / bilangan2}")
print(f"Hasil pembagian bulat (//): {bilangan1 // bilangan2}")
print(f"Hasil modulus (%)       : {bilangan1 % bilangan2}")
```

**Soal B2.4**

*Buatlah program Python yang menghitung efisiensi termal sebuah unit pembangkit menggunakan rumus:*

*efisiensi (%) = (energi output / energi input) * 100*

*Program meminta pengguna memasukkan nilai energi input dan energi output (dalam MJ), lalu menampilkan hasil efisiensi dengan f-string, dibulatkan 2 angka desimal. Pastikan penggunaan tanda kurung pada rumus sudah benar sesuai operator precedence.*



```python
# Nama      : Najwa Alfaniy
# NIM       : 4232501032
# Tanggal   : 23 September 2026
# Tujuan    : Menghitung efisiensi termal unit pembangkit.

energi_input = float(input("Masukkan energi input (MJ): "))
energi_output = float(input("Masukkan energi output (MJ): "))

efisiensi = (energi_output / energi_input) * 100

print(f"Efisiensi termal unit pembangkit adalah {efisiensi:.2f}%")
```
